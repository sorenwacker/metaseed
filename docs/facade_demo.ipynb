# Interactive Facade Demo

This notebook demonstrates the interactive facade for creating MIAPPE and ISA entities with tab completion and help functionality.

## 1. Getting Started

The facade provides a user-friendly interface that works well in Jupyter notebooks and IPython with:
- Tab completion for entity names and methods
- Built-in help for field information
- Example code generation
- Field search functionality

```python
from metaseed import miappe, isa, ProfileFacade, MetaseedClient
```

## 2. Creating a Profile Facade

Use the convenience functions `miappe()` and `isa()` or create a `ProfileFacade` directly.

```python
# Always use explicit profile and version
m = miappe(version="1.2")
print(m)

# Or using ProfileFacade directly with explicit args
m = ProfileFacade(profile="miappe", version="1.2")
print(m)
```

## 3. Discovering Entities

List all available entities in the profile.

```python
# List all entities
print("Available entities:")
for entity in m.entities:
    print(f"  - {entity}")
```

```python
# Profile overview with help()
m.help()
```

## 4. Entity Helpers

Access entity helpers via attribute access. In Jupyter/IPython, use Tab to see available entities.

```python
# Access an entity helper (use Tab completion: m.<Tab>)
inv = m.Investigation
print(inv)

# Case-insensitive access also works
inv2 = m.investigation
print(inv2)
```

## 5. Field Information

Get detailed information about entity fields.

```python
# Show all field information
m.Investigation.help()
```

```python
# Get required fields
print("Required fields:")
print(m.Investigation.required_fields)

print("\nOptional fields:")
print(m.Investigation.optional_fields)
```

```python
# Get detailed info for a specific field
print("unique_id field info:")
print(m.Investigation.field_info("unique_id"))

print("\ntitle field info:")
print(m.Investigation.field_info("title"))
```

```python
# See which fields contain nested entities
print("Nested entity fields:")
print(m.Investigation.nested_fields)
```

## 6. Example Code Generation

Generate example code for creating entities.

```python
# Generate example code
m.Investigation.example()
```

```python
m.Study.example()
```

## 7. Creating Entities

Create entity instances using the helper.

```python
import datetime

# Create an investigation using .create()
investigation = m.Investigation.create(
    unique_id="INV-001",
    title="Drought Stress Study in Wheat",
    description="Investigating drought response in wheat varieties",
    submission_date=datetime.date(2024, 1, 15),
)

print(f"Created: {investigation.title}")
print(f"ID: {investigation.unique_id}")
```

```python
# Or call the helper directly (shorthand)
# In MIAPPE 1.2, Study requires investigation_id reference
study = m.Study(
    unique_id="STU-001",
    investigation_id="INV-001",  # Reference to parent Investigation
    title="Field Trial 2024",
    description="Multi-location field trial",
)

print(f"Created: {study.title}")
print(f"References Investigation: {study.investigation_id}")
```

## 8. Search Functionality

Search for entities or fields by keyword.

```python
# Search for entities or fields containing "date"
results = m.search("date")
print("Fields containing 'date':")
for r in results:
    print(f"  {r}")
```

```python
# Search for "sample"
results = m.search("sample")
print("Results for 'sample':")
for r in results:
    print(f"  {r}")
```

## 9. Working with ISA Profile

The same facade works with ISA profile entities.

```python
# Create ISA facade
i = isa()
print(i)

print("\nISA entities:")
for entity in i.entities:
    print(f"  - {entity}")
```

```python
# Create ISA entities
isa_inv = i.Investigation(
    identifier="MTBLS-001",
    title="Metabolomics Study",
    description="LC-MS metabolite profiling",
)

print(f"ISA Investigation: {isa_inv.title}")
```

## 10. Working with MetaboLights Profile

MetaboLights uses a unified Assay entity with an `assay_type` discriminator to handle NMR, LC-MS, and GC-MS experiments. A study can combine multiple techniques.

```python
from metaseed import metabolights

# Create MetaboLights facade
mtbls = metabolights()
print(mtbls)

print("\nMetaboLights entities:")
for entity in mtbls.entities:
    print(f"  - {entity}")

# Show Assay fields
print("\nAssay fields:")
mtbls.Assay.help()
```

```python
# Create a MetaboLights Assay with correct fields
assay = mtbls.Assay(
    identifier="ASSAY-001",
    filename="a_MTBLS001_metabolite_profiling.txt",
    technology_type="mass spectrometry",
    measurement_type="metabolite profiling",
    technology_platform="Thermo Q Exactive Plus",
    samples=["Sample1", "Sample2", "Sample3"],
)

print(f"Created Assay: {assay.identifier}")
print(f"  Technology: {assay.technology_type}")
print(f"  Samples: {assay.samples}")
```

## 11. Building Entity Collections

In MIAPPE 1.2, entities are linked via ID references rather than nesting. Create collections of related entities.

```python
# Create an investigation and related studies
investigation = m.Investigation(
    unique_id="INV-COMPLETE",
    title="Complete Phenotyping Project",
    description="A comprehensive phenotyping investigation",
)

# Studies reference the investigation via investigation_id
studies = [
    m.Study(
        unique_id="STU-001",
        investigation_id="INV-COMPLETE",
        title="Greenhouse Trial",
        description="Controlled environment study",
    ),
    m.Study(
        unique_id="STU-002",
        investigation_id="INV-COMPLETE",
        title="Field Trial",
        description="Multi-location field experiment",
    ),
]

print(f"Investigation: {investigation.title}")
print(f"Studies referencing {investigation.unique_id}:")
for study in studies:
    print(f"  - {study.title} (investigation_id: {study.investigation_id})")
```

## 12. MetaseedClient vs ProfileFacade

Both APIs are fully supported. Choose based on your use case:

| Feature | ProfileFacade | MetaseedClient |
|---------|---------------|----------------|
| **Primary use** | Interactive (Jupyter/IPython) | Programmatic (scripts, apps) |
| **Tab completion** | Yes | No |
| **Direct model creation** | `facade.Investigation(...)` | `client.create_entity("Investigation", {...})` |
| **Built-in help** | `entity.help()`, `entity.example()` | Schema introspection methods |
| **Entity storage** | Via `add_entity()` | Built-in with CRUD operations |
| **Validation** | Manual via validators | `validate()`, `validate_entity()` |
| **Serialization** | `to_dict()`, `load_from_dict()` | `serialize()`, `load()` with format options |

**Recommendation:**
- Use **ProfileFacade** in Jupyter notebooks for exploration and interactive work
- Use **MetaseedClient** in scripts, web apps, and CLIs for clean programmatic access

```python
# MetaseedClient example for comparison
client = MetaseedClient("miappe", "1.2")

# Create entities programmatically
inv = client.create_entity(
    "Investigation",
    {"unique_id": "INV-PROG-001", "title": "Programmatic Investigation"},
)

# Validation and serialization built-in
result = client.validate()
data = client.serialize(format="tree")

print(f"Created: {inv.entity_type} - {inv.label}")
print(f"Valid: {result.valid}")
print(f"Serialized {len(data['tree'])} root entities")
```

## Summary

The interactive facade provides:

1. **Tab completion**: Access entities with `profile.<Tab>`
2. **Help system**: `entity.help()` for field documentation
3. **Field discovery**: `required_fields`, `optional_fields`, `nested_fields`
4. **Example generation**: `entity.example()` for code templates
5. **Search**: `profile.search("keyword")` to find entities/fields
6. **Creation**: `entity.create()` or `entity()` to instantiate

**MIAPPE 1.2**: Uses flat references (`investigation_id`, `study_id`) rather than nesting.

Shortcut functions available:
- `miappe()` - MIAPPE plant phenotyping
- `isa()` - ISA framework  
- `metabolights()` - MetaboLights metabolomics
- `ena()` - ENA nucleotide archive
- `pride()` - PRIDE proteomics
- `dissco()` - DiSSCo specimen data
- `darwin_core()` - Darwin Core biodiversity

For programmatic use (scripts, web apps), use `MetaseedClient` instead.
